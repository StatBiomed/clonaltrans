# Filter LARRY Barcodes with length 40; and 10x CBC UMI with length 28

```python
larry_unfiltered=np.loadtxt("larrybarcode_file_D17Mid.txt", dtype=str)
```

```python
r1=np.loadtxt("10xcbcUMI_file_D17Mid.txt", dtype=str)
```

```python
length_checker = np.vectorize(len)
larry_barcode_len = length_checker(larry_unfiltered)
```

```python
mask_1 = np.where(larry_barcode_len == 40)
```

```python
larry_filtered = larry_unfiltered[mask_1]
```

```python
r1_filtered = r1[mask_1]
```

```python
r1_len = length_checker(r1_filtered)
```

```python
mask_2 = np.where(r1_len == 28)
```

```python
r1_filtered = r1_filtered[mask_2]
```

```python
larry_filtered = larry_filtered[mask_2]
```

# Extracting 10x Barcode and UMI

```python
cbc = []
umi = []
for i in r1_filtered:
    x = i[:16]
    y = i[16:]
    cbc.append(x)
    umi.append(y)
```

```python
cbc = np.array(cbc)
```

```python
umi = np.array(umi)
```

# Generate fastq file with correct configuration

```python
out = gzip.open('LARRY_sorted_and_filtered_barcodes_D17Mid.fastq.gz','wb')
```

```python
for i in range(len(larry_filtered)):
    lib = 'SITTB9'
    gfp_bc = larry_filtered[i]
    umi = umi_filtered[i]
    cell_bc = cbc_filtered[i]
    tag = '>'+lib+','+cell_bc+','+umi
    out.write((tag+'\n').encode('utf-8'))
    out.write((gfp_bc+'\n').encode('utf-8'))
    out.write('\n'.encode('utf-8'))
```

```python
out.close()
```
