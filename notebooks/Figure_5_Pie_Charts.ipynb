```python
adata = sc.read_h5ad('../datasets/Mouse_Hematopoiesis/adata_used.h5ad')
adata
```

```python
adata.var_names_make_unique()
adata = adata[adata.obs['meta_clones'] != '-1.0', :]
idx_mt = [not gene.startswith('mt-') for gene in adata.var.index]
adata = adata[:, idx_mt]
adata
```

```python
labels = ['early_prog_Neu', 'prog_Neu', 'prog_Mono', 'Mono']

patches, texts = plt.pie(
    [11, 0, 13, 3], colors=['#e6afb9', '#f3e1eb', '#d5eae7','#8595e1'], 
    wedgeprops = {
        "edgecolor" : "white", 
        'linewidth': 2, 
        'antialiased': True
    }, startangle=90, counterclock=False)
plt.title('Meta-clone 5 day 6')

plt.savefig("pie_chart_mc5_day6_mouse.svg", bbox_inches='tight', dpi=300, transparent=True)
```
